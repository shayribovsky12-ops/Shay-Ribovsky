```python
import os
with open('data2.csv', 'w') as f:
    f.write('x, y\n')
    for val in range(101):
        f.write(f'{val}, {val**2}\n')

os.startfile('data2.csv')
```

```python
with open('data2.csv', 'r') as f:
    content = f.read()
print(content)
```

```python
import matplotlib.pyplot as plt

x_values = []
y_values = []

with open('data2.csv', 'r') as f:
    next(f)
    for line in f:
        parts = line.split(',')
        x_values.append(int(parts[0]))
        y_values.append(int(parts[1]))

fig, ax = plt.subplots()
ax.plot(x_values, y_values, linestyle='--')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y = x^2')
plt.show()
```
